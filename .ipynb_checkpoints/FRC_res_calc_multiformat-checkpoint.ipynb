# FRC resolution estimation: image folders, HDF5, and imzML

Estimate single-image FRC resolution across channels/peaks. Inputs can be a folder of image files, an HDF5 data cube, or an imzML MSI dataset. Peaks are ranked by mean image intensity; top channels can be transferred between comparable datasets using stable channel indices.


## Dependencies and configuration

For image folders, install `imageio`. For HDF5, install `h5py`. For imzML, install `pyimzML` (`pip install pyimzML`). MIPLIB must be available in the Python environment.


```python
from pathlib import Path
import re

import imageio.v2 as imageio
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

try:
    import h5py
except ImportError:
    h5py = None

try:
    from pyimzml.ImzMLParser import ImzMLParser
except ImportError:
    ImzMLParser = None

from miplib.analysis.resolution import analysis as fscanalysis
from miplib.analysis.resolution import fourierringcorrelation
from miplib.data.containers.fouriercorrelationdata import FourierCorrelationDataCollection
from miplib.data.containers.image import Image
from miplib.data.iterators.fourierringiterators import FourierRingIterator
from miplib.processing import image as imops
from miplib.processing import windowing
from miplib.ui.cli import miplibentrypointoptions as options

# INPUT_TYPE: 'folder', 'h5', or 'imzml'
INPUT_TYPE = 'folder'
INPUT_PATH = Path(r'E:/HyReS/results/DS brains/GT h5/brains4_top100')
OUTPUT_DIR = INPUT_PATH if INPUT_TYPE == 'folder' else INPUT_PATH.parent
OUTPUT_CSV = OUTPUT_DIR / 'FRC_resolutions.csv'
SELECTION_CSV = OUTPUT_DIR / 'FRC_peak_selection.csv'

# Shared settings
PIXEL_SIZE_UM = 100.0
TOP_N = 100                       # None evaluates all channels
REFERENCE_SELECTION_CSV = None     # Transfer a reference channel-index list
STRICT_CHANNEL_MATCH = True
APPLY_HAMMING = True
AVERAGE_SPLITS = True

# Folder settings
IMAGE_EXTENSIONS = {'.tif', '.tiff', '.png'}
CHANNEL = None                     # Set only for multichannel image files

# HDF5 settings
H5_DATASET_KEY = 'data'            # Dataset path, e.g. 'data', 'cube', or 'images'
H5_CHANNEL_AXIS = 0                # Axis corresponding to spectral channels
H5_MZ_DATASET_KEY = None           # Optional 1D m/z dataset key, e.g. 'mz'

# imzML settings
IMZML_BIN_WIDTH = 0.01             # m/z bin width used to form ion images
IMZML_MZ_RANGE = None              # None, or tuple e.g. (100.0, 1000.0)

args_list = ('None --bin-delta=1 --frc-curve-fit-type=polynomial '
             '--resolution-threshold-criterion=fixed --resolution-threshold-value=0.5').split()
args = options.get_frc_script_options(args_list)
```

## Input loaders

All loaders return a `PeakSource` object. Its `metadata` table contains stable `channel_index` values, labels/m/z where available, and mean image intensity. The actual 2D image is constructed only when needed for FRC.


```python
def natural_sort_key(path):
    return [int(token) if token.isdigit() else token.lower()
            for token in re.split(r"(\d+)", path.name)]


class PeakSource:
    def __init__(self, metadata, get_image):
        self.metadata = metadata
        self.get_image = get_image


def read_file_image(path):
    image = np.asarray(imageio.imread(path), dtype=np.float64)
    if image.ndim == 3:
        if CHANNEL is None:
            raise ValueError(f'Multichannel image requires CHANNEL: {path.name}')
        image = image[..., CHANNEL]
    if image.ndim != 2:
        raise ValueError(f'Expected 2D image, got {image.shape}: {path.name}')
    return image


def load_folder_source(directory):
    paths = sorted(
        (p for p in Path(directory).iterdir() if p.suffix.lower() in IMAGE_EXTENSIONS),
        key=natural_sort_key,
    )
    if not paths:
        raise FileNotFoundError(f'No supported images found in {directory}')
    records = []
    for index, path in enumerate(paths):
        try:
            image = read_file_image(path)
            records.append({'channel_index': index, 'label': path.stem, 'filename': path.name,
                            'mean_intensity': float(image.mean()), 'max_intensity': float(image.max()),
                            'n_pixels': int(image.size), 'metadata_error': ''})
        except Exception as exc:
            records.append({'channel_index': index, 'label': path.stem, 'filename': path.name,
                            'mean_intensity': np.nan, 'max_intensity': np.nan,
                            'n_pixels': np.nan, 'metadata_error': str(exc)})
    return PeakSource(pd.DataFrame(records), lambda index: read_file_image(paths[int(index)]))


def _move_channel_axis_to_front(array, channel_axis):
    return np.moveaxis(array, channel_axis, 0)


def load_h5_source(path):
    if h5py is None:
        raise ImportError('h5py is required for HDF5 input. Install with: pip install h5py')
    path = Path(path)
    with h5py.File(path, 'r') as handle:
        if H5_DATASET_KEY not in handle:
            raise KeyError(f"H5_DATASET_KEY '{H5_DATASET_KEY}' not found. Available keys: {list(handle.keys())}")
        shape = handle[H5_DATASET_KEY].shape
        mz_values = None
        if H5_MZ_DATASET_KEY is not None:
            mz_values = np.asarray(handle[H5_MZ_DATASET_KEY])
    if len(shape) != 3:
        raise ValueError(f'HDF5 data cube must be 3D; got shape {shape}.')
    n_channels = shape[H5_CHANNEL_AXIS]
    if mz_values is not None and len(mz_values) != n_channels:
        raise ValueError('Length of H5 m/z vector does not match the channel axis.')

    def get_image(index):
        with h5py.File(path, 'r') as handle:
            data = handle[H5_DATASET_KEY]
            slicer = [slice(None)] * 3
            slicer[H5_CHANNEL_AXIS] = int(index)
            return np.asarray(data[tuple(slicer)], dtype=np.float64)

    records = []
    for index in range(n_channels):
        try:
            image = get_image(index)
            label = f'm/z {mz_values[index]:.4f}' if mz_values is not None else f'channel_{index}'
            records.append({'channel_index': index, 'label': label,
                            'mz': float(mz_values[index]) if mz_values is not None else np.nan,
                            'mean_intensity': float(image.mean()), 'max_intensity': float(image.max()),
                            'n_pixels': int(image.size), 'metadata_error': ''})
        except Exception as exc:
            records.append({'channel_index': index, 'label': f'channel_{index}', 'mz': np.nan,
                            'mean_intensity': np.nan, 'max_intensity': np.nan,
                            'n_pixels': np.nan, 'metadata_error': str(exc)})
    return PeakSource(pd.DataFrame(records), get_image)


def load_imzml_source(path):
    if ImzMLParser is None:
        raise ImportError('pyimzML is required for imzML input. Install with: pip install pyimzML')
    parser = ImzMLParser(str(path))
    coordinates = np.asarray(parser.coordinates)
    xy = coordinates[:, :2].astype(int)
    x0, y0 = xy.min(axis=0)
    width, height = xy.max(axis=0) - [x0, y0] + 1

    mz_min = np.inf
    mz_max = -np.inf
    for spectrum_index in range(len(parser.coordinates)):
        mzs, _ = parser.getspectrum(spectrum_index)
        if len(mzs):
            mz_min = min(mz_min, float(np.min(mzs)))
            mz_max = max(mz_max, float(np.max(mzs)))
    if IMZML_MZ_RANGE is not None:
        mz_min, mz_max = IMZML_MZ_RANGE
    edges = np.arange(mz_min, mz_max + IMZML_BIN_WIDTH, IMZML_BIN_WIDTH)
    centers = (edges[:-1] + edges[1:]) / 2

    def get_image(index):
        lo, hi = edges[int(index)], edges[int(index) + 1]
        image = np.zeros((height, width), dtype=np.float64)
        for spectrum_index, (x, y, *_rest) in enumerate(parser.coordinates):
            mzs, intensities = parser.getspectrum(spectrum_index)
            mask = (mzs >= lo) & (mzs < hi)
            image[int(y) - y0, int(x) - x0] = np.sum(intensities[mask])
        return image

    records = []
    for index, mz in enumerate(centers):
        try:
            image = get_image(index)
            records.append({'channel_index': index, 'label': f'm/z {mz:.4f}', 'mz': float(mz),
                            'mean_intensity': float(image.mean()), 'max_intensity': float(image.max()),
                            'n_pixels': int(image.size), 'metadata_error': ''})
        except Exception as exc:
            records.append({'channel_index': index, 'label': f'm/z {mz:.4f}', 'mz': float(mz),
                            'mean_intensity': np.nan, 'max_intensity': np.nan,
                            'n_pixels': np.nan, 'metadata_error': str(exc)})
    return PeakSource(pd.DataFrame(records), get_image)


def load_peak_source():
    if INPUT_TYPE == 'folder':
        return load_folder_source(INPUT_PATH)
    if INPUT_TYPE == 'h5':
        return load_h5_source(INPUT_PATH)
    if INPUT_TYPE == 'imzml':
        return load_imzml_source(INPUT_PATH)
    raise ValueError("INPUT_TYPE must be 'folder', 'h5', or 'imzml'.")
```

## Selection and FRC functions

For fair comparisons, create `FRC_peak_selection.csv` from a reference dataset, then set `REFERENCE_SELECTION_CSV` for the target dataset. This transfers channel indices; for HDF5/imzML, the channel/bin ordering must be identical between datasets.


```python
def select_peaks(peaks, top_n=None, reference_selection_csv=None, strict_channel_match=True):
    peaks = peaks.sort_values('channel_index').reset_index(drop=True).copy()
    if reference_selection_csv is not None:
        reference = pd.read_csv(reference_selection_csv)
        if 'channel_index' not in reference.columns:
            raise ValueError("Reference selection CSV must contain 'channel_index'.")
        reference['channel_index'] = reference['channel_index'].astype(int)
        available = set(peaks.channel_index)
        missing = sorted(set(reference.channel_index) - available)
        if missing:
            message = f'Missing reference indices: {missing[:10]}' + ('...' if len(missing) > 10 else '')
            if strict_channel_match:
                raise IndexError(message)
            print('Warning:', message)
            reference = reference[reference.channel_index.isin(available)]
        selected = reference.merge(peaks, on='channel_index', how='left', suffixes=('_reference', ''), validate='one_to_one')
        selected['reference_channel_index'] = selected['channel_index']
        selected['selection_source'] = 'reference_channel_index'
        return selected.reset_index(drop=True)

    ranked = peaks.dropna(subset=['mean_intensity']).sort_values('mean_intensity', ascending=False, kind='stable').reset_index(drop=True)
    ranked['intensity_rank'] = np.arange(1, len(ranked) + 1)
    ranked['selection_source'] = 'all_channels' if top_n is None else f'top_{top_n}_mean_intensity'
    if top_n is None:
        return ranked
    if not isinstance(top_n, int) or isinstance(top_n, bool) or top_n < 1:
        raise ValueError('TOP_N must be a positive integer or None.')
    return ranked.head(top_n).reset_index(drop=True)


def estimate_frc_resolution(array):
    image = Image(array, PIXEL_SIZE_UM, PIXEL_SIZE_UM)
    if APPLY_HAMMING and not args.disable_hamming:
        image = Image(windowing.apply_hamming_window(image), image.spacing)
    image = imops.zero_pad_to_cube(image)
    image1, image2 = imops.checkerboard_split(image)
    image1, image2 = imops.zero_pad_to_matching_shape(image1, image2)
    frc_data = FourierCorrelationDataCollection()
    task = fourierringcorrelation.FRC(image1, image2, FourierRingIterator(image1.shape, args.dbin), frc_data[0])
    task.execute()
    if AVERAGE_SPLITS:
        image1, image2 = imops.reverse_checkerboard_split(image)
        image1, image2 = imops.zero_pad_to_matching_shape(image1, image2)
        task = fourierringcorrelation.FRC(image1, image2, FourierRingIterator(image1.shape, args.dbin), frc_data[0])
        frc_data[0].correlation['correlation'] *= 0.5
        frc_data[0].correlation['correlation'] += 0.5 * task.execute().correlation['correlation']
    results = fscanalysis.FourierCorrelationAnalysis(frc_data, image.spacing[0], args).execute(z_correction=0)
    result = results[0]
    point = result.resolution.get('resolution-point')
    return float(result.resolution.get('resolution')) if result.resolution.get('resolution') is not None else np.nan, float(point[1]) if point is not None else np.nan
```

## Load, rank and select peaks


```python
source = load_peak_source()
all_peaks = source.metadata
selected_peaks = select_peaks(all_peaks, TOP_N, REFERENCE_SELECTION_CSV, STRICT_CHANNEL_MATCH)
selected_peaks.to_csv(SELECTION_CSV, index=False)
print(f'Input type: {INPUT_TYPE}')
print(f'Available channels: {len(all_peaks)}')
print(f'Selected channels: {len(selected_peaks)}')
print(f'Selection CSV: {SELECTION_CSV}')
selected_peaks[['channel_index', 'label', 'mean_intensity', 'intensity_rank']].head(10)
```

## Evaluate FRC


```python
records = []
for _, peak in selected_peaks.iterrows():
    try:
        image = source.get_image(int(peak.channel_index))
        resolution_um, cutoff_frequency = estimate_frc_resolution(image)
        records.append({**peak.to_dict(), 'cutoff_frequency': cutoff_frequency,
                        'estimated_resolution_um': resolution_um, 'status': 'ok', 'error': ''})
    except Exception as exc:
        records.append({**peak.to_dict(), 'cutoff_frequency': np.nan,
                        'estimated_resolution_um': np.nan, 'status': 'failed', 'error': str(exc)})
results = pd.DataFrame(records)
results.to_csv(OUTPUT_CSV, index=False)
results.head()
```

## Summary


```python
valid = results.query("status == 'ok'").dropna(subset=['estimated_resolution_um'])
print(f'Successful estimates: {len(valid)}/{len(results)}')
if len(valid):
    print(f'Mean resolution: {valid.estimated_resolution_um.mean():.2f} µm')
    print(f'Median resolution: {valid.estimated_resolution_um.median():.2f} µm')
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(valid.estimated_resolution_um, bins=30)
axes[0].set(xlabel='Estimated resolution (µm)', ylabel='Channel count')
axes[1].scatter(valid.mean_intensity, valid.estimated_resolution_um, s=18, alpha=0.7)
axes[1].set(xlabel='Mean image intensity', ylabel='Estimated resolution (µm)')
if len(valid) and (valid.mean_intensity > 0).all(): axes[1].set_xscale('log')
plt.tight_layout()
```
